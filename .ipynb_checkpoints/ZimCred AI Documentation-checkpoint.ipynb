# ZimCred AI – Integration Blueprint  
## Version 1.0 – Institutional API Integration Guide

---

## 1. Overview

ZimCred AI is a B2B alternative credit risk engine designed to integrate directly into an MFI’s existing loan management system.

It provides:

- Probability of Default (PD)
- Credit Score (300–850 scale)
- Risk Band classification
- Approved Loan Limit
- Final Decision
- Persistent decision logging (bureau layer)

ZimCred AI does not replace an MFI’s loan system.  
It acts as a risk intelligence layer.

---

## 2. System Flow

Borrower Applies  
→ MFI collects application data  
→ MFI backend sends JSON to ZimCred `/score` endpoint  
→ ZimCred evaluates risk  
→ ZimCred returns structured decision  
→ Decision logged in bureau database  
→ MFI proceeds with internal approval process  

---

## 3. API Endpoint

### POST `/score`

### Request Format

```json
{
  "monthly_income": 500,
  "requested_loan_amount": 300,
  "monthly_expenses": 200,
  "dependents": 2
}
```

---

### Response Format

```json
{
  "Probability_of_Default": 0.3824,
  "Credit_Score": 639,
  "Risk_Band": "Medium Risk",
  "Recommended_Loan_Limit": 45000
}
```

---

## 4. Bureau Logging

Each scoring request is securely stored in the ZimCred decision database, including:

- Timestamp
- Applicant financial inputs
- Probability of Default
- Credit Score
- Risk Band
- Loan decision outcome

This creates structured alternative credit history over time.

---

## 5. Multi-MFI Policy Layer

ZimCred supports multiple MFIs using a shared ML engine.

Each MFI can configure:

- Risk thresholds
- Loan caps per risk band
- Approval aggressiveness

Core ML model remains centralized.  
Institution policies remain customizable.

---

## 6. Technical Requirements for MFIs

To integrate, an MFI must have:

- A backend system capable of sending HTTP POST requests
- Basic internet connectivity
- Ability to process JSON responses

No advanced AI infrastructure required.

---

## 7. Security & Future Enhancements

Upcoming features:

- API key authentication
- Rate limiting
- Encrypted communication
- PostgreSQL migration for scale
- Cloud deployment
- AI Assistance Layer for loan officers

---

## 8. Version Status

ZimCred AI v1 includes:

✔ ML risk engine  
✔ Alternative behavioral scoring  
✔ Multi-MFI policy layer  
✔ SQLite bureau logging  
✔ FastAPI integration layer

# ZimCred AI – Help Center Agent Architecture (Beginner-Friendly Explanation)

---

# 1. First: What Is The AI Help Center Agent?

It is NOT the credit scoring model.

It is NOT making loan decisions.

It is an intelligent assistant that:

- Explains ZimCred decisions
- Explains MFI policies
- Guides loan officers
- Helps MFI staff understand risk bands
- Assists with borrower questions (through the MFI)

It does NOT replace the MFI.
It assists the MFI.

Think of it as:

A smart internal assistant connected to your scoring engine.

---

# 2. What Type of AI Is This?

This is Generative AI (LLM-based).

Unlike your ML risk model (XGBoost),
this uses a Large Language Model (like GPT-style models).

Your ML model calculates:
→ Probability of Default

Your AI Agent explains:
→ Why that decision happened
→ What it means
→ What to do next

---

# 3. How The Architecture Works (Simple Version)

There are 4 parts:

1️⃣ User (MFI staff member)  
2️⃣ AI Agent  
3️⃣ ZimCred Scoring API  
4️⃣ Knowledge Base (MFI policies + rules)

Flow:

MFI Staff asks:
"Why was this applicant declined?"

↓

AI Agent retrieves:
- Decision from ZimCred API
- Risk band
- SHAP drivers
- MFI policy thresholds

↓

AI Agent explains in plain language:
"This applicant was classified as High Risk because income volatility was high and debt ratio exceeded MFI_B threshold."

That’s it.

The agent does not calculate risk.
It interprets and explains.

---

# 4. Where Generative AI Fits

Generative AI is used to:

- Convert raw risk outputs into human explanations
- Adjust tone based on institution
- Translate technical risk into business language
- Help staff make consistent decisions

Without it:
You return JSON numbers.

With it:
You return intelligence.

---

# 5. Technical Architecture (Practical Layer)

ZimCred System:

[Scoring Engine] → produces decision

[Policy Engine] → applies MFI rules

[Database] → stores decisions

[AI Agent Layer] → sits on top

The AI Agent:

- Pulls decision data
- Pulls policy configuration
- Uses prompt templates
- Sends structured context to LLM
- Returns structured explanation

Important:
The agent NEVER makes loan decisions.
It only interprets them.

---

# 6. Two Modes of Operation

Mode A – Internal MFI Assistant
Used by loan officers.

Mode B – Customer Education Assistant
Explains general risk logic without exposing sensitive details.

For now:
We build Mode A only.

Keep it simple.

---

# 7. What We Need To Build It

You already have:

✔ Scoring API  
✔ Risk bands  
✔ SHAP feature drivers  
✔ Policy layer  
✔ Bureau logs  

To build the AI Agent we need:

That’s it.

No retraining needed.
No new ML model needed.

# 8. Important Beginner Principle

Your ML model answers:

"What is the risk?"

Your LLM agent answers:

They are different systems.

Do not confuse them.

# 9. Why This Is Powerful

MFIs struggle with:

# ZimCred AI – System Overview & Architecture

## 1. Introduction

ZimCred AI is an AI-powered loan decision support system designed to help micro-lenders make faster and smarter credit decisions.

The system collects borrower information, processes it through a machine learning model, and produces a risk score that helps determine whether a loan should be approved or declined.

The goal is not to replace human lenders, but to **assist decision making with data-driven insights**.

---

# 2. Core Idea

Traditional micro-lending often relies on manual judgement, which can be:

- Slow
- Inconsistent
- Prone to bias
- Difficult to scale

ZimCred AI improves this process by:

1. Collecting borrower data
2. Running the data through a trained AI model
3. Generating a credit risk score
4. Presenting a recommendation to the lender

---

# 3. System Components

The system is made up of four main components.

## 3.1 Frontend Interface

This is the **user interface** where the lender interacts with the system.

Example tools used:
- Streamlit
- Simple Web Forms

Functions:
- Capture borrower details
- Display loan prediction results
- Show decision history

---

## 3.2 Backend API

The backend handles communication between the interface and the AI model.

Technology used:
- FastAPI

Responsibilities:
- Receive loan application data
- Send data to the AI model
- Return prediction results
- Store decision history

---

## 3.3 Machine Learning Model

The machine learning model is the **core intelligence of the system**.

Libraries used:
- Scikit-learn
- Pandas
- NumPy

The model analyzes borrower features such as:

- Income
- Loan amount
- Repayment history
- Employment stability
- Other financial indicators

It then produces a **risk probability score**.

Example Output:

Risk Score: 0.18  
Interpretation: Low risk borrower.

---

## 3.4 Data Storage

The system stores loan decisions and application data.

Possible storage options:

- CSV files (for prototype stage)
- SQLite
- PostgreSQL

Stored data can include:

- Applicant details
- Loan amount
- Model prediction
- Final decision
- Timestamp

---

# 4. System Workflow

The full workflow of ZimCred AI looks like this:

1. Lender opens the application interface.
2. Lender enters borrower information.
3. The frontend sends the data to the backend API.
4. The API processes the request.
5. The machine learning model evaluates the borrower.
6. The model returns a risk score.
7. The system displays the recommendation.
8. The decision is saved in the database.

---

# 5. Current Prototype Capabilities

The current prototype can:

- Accept loan application inputs
- Generate AI credit risk predictions
- Display results in real time
- Store basic decision history

This demonstrates the **core concept of AI-assisted lending**.

---

# 6. Planned Features

Future improvements may include:

- WhatsApp loan application system
- Automated borrower verification
- More advanced risk models
- Fraud detection
- Credit scoring dashboards
- Integration with financial institutions

---

# 7. Ethical and Legal Considerations

The system includes a disclaimer because:

- AI recommendations are **not final decisions**
- Human oversight is required
- Lending decisions must comply with financial regulations

The system is designed as a **decision support tool**, not a fully autonomous lender.

---

# 8. Conclusion

ZimCred AI demonstrates how artificial intelligence can assist micro-lending operations by improving speed, consistency, and scalability.

By combining:

- Machine learning
- API infrastructure
- Simple user interfaces

The system provides a practical foundation for building modern AI-powered lending platforms.
